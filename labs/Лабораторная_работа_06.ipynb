Теперь ответим на поставленные вопросы, используя строку `mystr`

```python
mystr = open('pr6.txt').read().strip()
# mystr='ASD.DFGUIO.OOOAEEERUY.UYTEE'
glas = 'AEIOUY'

# заменим гласные на символ *
for c in glas:
  mystr = mystr.replace(c, '*')

# порежем строку по символу *
sub_str = mystr.split('.')
max_len = 0

# выполним обход всех подстрок ...
for s in sub_str:
  z = s.split('*')
  # ... и проанализируем каждую подстроку
  for i in range(len(z)-7):
    t = ''.join(z[i:i+8])
    if max_len < len(t):
      max_len = len(t) + len(z[i:i+8])-1
      print(t)
      print(z[i:i+8])

print(f'Максимальная подходящая подстрока: {max_len}')
```

```python
# Код для решения индивидуального задания

mystr = open('pr6_1.txt').read().strip()
count = 0

for i in range(len(mystr)):
    # является ли символ цифрой
    if mystr[i].isdigit():
        #разделяем строки
        left = mystr[:i]
        right = mystr[i+1:]
        #проверяем условие
        if any(c.isalpha() for c in left) and any(c.isdigit() for c in right):
            count += 1

print('Кол-во цифр:', count)

```
