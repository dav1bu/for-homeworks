На данном этапе проектирование завершается. Переходим к объявлению класса `Par`.

```python
class Par:
  # Объявляем конструктор класса
  def __init__(self, pa, pb, pc):
    self.a=pa
    self.b=pb
    self.c=pc

  # Объявляем метод для вычисления площади поверхности фигуры
  def GetS(self):
    return 2*(self.a*self.b + self.b*self.c + self.a*self.c)

  # Объявляем метод для вычисления объема
  def GetV(self):
    return self.a*self.b*self.c

  # Объявление метода для вывода информации о фигуре
  def Info(self):
    print('*'*20)
    print('Прямоугольный параллелепипед: \n')
    print('Размеры: %.2f x %.2f x %.2f' % (self.a, self.b, self.c))
    print('Площадь: %.3f' % self.GetS())
    print('Объем: %.3f' % self.GetV())
    print('*'*20)

  def Load(path_to_file):
    # Предполагаем, что в файле записаны 3 числа в строку через пробел
    d = open(path_to_file).read().strip().split()
    tmp = Par(float(d[0]), float(d[0]), float(d[0]))
    return tmp
```

В данном листинге производится объявление типа данных `Par` с использованием зарезервированного слова `class`. Внутри класса объявляются функции (методы класса) и переменные (поля данных класса). Следует обратить внимание на использование зарезервированного слова `self`.
Также внутри класса объявлен метод `__init__()`. Данный метод имеет специальное назначение - он вызывается в момент создания переменной данного класса автоматически. Данный метод называется конструктор класса.

Использование разработанного типа данных `Par`

```python
# Создание объекта p1 типа Par (используется конструктор)
p1 = Par(10, 20, 30)
# Вызов метода класса
p1.Info()
```

Следует обратит внимание на объявление метода `Load()` в классе Par. Несмотря на его принадлежность классу, сам метод не содержит параметра `self`.

Это значит, что метод не принадлежит конкретному экземпляру (объекту), он принадлежит самому классу. Такие методы называют статическими. Для вызова статического метода нет необходимости явно создавать объект класса, используя конструктор. Статический метод вызывается из класса.

# Индивидуальное задание

Код задания

```python
import math

class Cylinder:
    # Объявляем конструктор класса
    def __init__(self, pr, ph):
        self.__r = pr
        self.__h = ph

    # Метод для ввода данных
    def Input(self):
        self.__r = float(input("Радиус цилиндра: "))
        self.__h = float(input("Высота цилиндра: "))

    # Метод для вычисления площади поверхности
    def GetS(self):
        return 2 * math.pi * self.__r * (self.__r + self.__h)

    # Метод для вычисления объема
    def GetV(self):
        return math.pi * self.__r ** 2 * self.__h

    # Метод для вывода информации о цилиндре
    def Info(self):
        print("*" * 30)
        print("Цилиндр:")
        print("Радиус: %.2f" % self.__r)
        print("Высота: %.2f" % self.__h)
        print("Площадь поверхности: %.3f" % self.GetS())
        print("Объем: %.3f" % self.GetV())
        print("*" * 30)

    # Статический метод для загрузки данных из файла
    @staticmethod
    def Load(path_to_file):
        # Предполагаем, что в файле записаны два числа через пробел
        d = open(path_to_file).read().strip().split()

        tmp = Cylinder(float(d[0]), float(d[1]))
        return tmp
```

Использование

```python
# Создание объекта c1 типа Cylinder
c1 = Cylinder(5, 10)

# Вызов метода
c1.Info()
```

### 4.  Для чего используется параметр self в каждом методе класса?
self — ссылка на текущий экземпляр класса. Используется для обращения к его атрибутам и другим методам.


#### 5. Сколько методов __init__ может быть объявлено внутри класса?
В классе может быть объявлен только один метод __init__(). При повторном объявлении предыдущая реализация будет заменена.

#### 6. Каким образом в классе Python атрибут скрывается от программного кода, находящегося за пределами класса?
Для сокрытия атрибута используют два символа подчёркивания перед его именем: self.__name. Python применяет механизм преобразования имени атрибута (name mangling).

#### 7. Какова задача метода __str__()?
Предназначен для определения удобного текстового представления объекта, которое выводится пользователю.

#### 8. Каким образом происходит вызов метода __str__()?
Вызывается автоматически при использовании функций print() и str() для объекта.

#### 9. Что такое атрибут экземпляра?
Это переменная, принадлежащая конкретному объекту класса и хранящая его индивидуальные данные.

#### 10. Что такое метод-получатель? Что такое метод-мутатор?
Метод-получатель (getter) — метод для получения значения атрибута. Метод-мутатор (setter) — метод для изменения значения атрибута.
