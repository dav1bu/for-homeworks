## Контрольные вопросы

### 1. Что такое переменная в Python?

Переменная — это имя, которому присваивается определённое значение.Присваивание выполняется с помощью =.

Примеры:


name = "Максим"age = 21

Некорректное имя:

### 2. Основные типы данных Python

- int — целые числа: 10 -> <class 'int'>
- float — дробные числа: 3.14 -> <class 'float'>
- str — строки: "Привет" -> <class 'str'>
- bool — логические значения: True -> <class 'bool'>

Тип динамический, потому что переменной можно присвоить значение другого типа.

### 3. Как работает print()?

Функция print() выводит информацию на экран.


name = "Максим"print(name)

print("Возраст:", 21)

print(f"Меня зовут {name}")


Первый способ выводит переменную, второй — несколько значений, третий — форматированную строку.

### 4. Что делает type()?

Функция type() определяет тип данных.


a = 10b = "Привет"c = True

print(type(a))

print(type(b))

print(type(c))


Результат: int, str, bool. Он показывает, к какому типу относится значение.

### 5. Арифметические операции Python

сложение: 5 + 2 = 7 (int)

вычитание: 5 - 2 = 3 (int)

умножение: 5 * 2 = 10 (int)

деление: 5 / 2 = 2.5 (float)

целочисленное деление: 5 // 2 = 2 (int)

остаток от деления: 5 % 2 = 1 (int)

возведение в степень: 5 ** 2 = 25 (int)

Если в вычислении участвует float, результат обычно тоже будет float.

### 6. Разница между / и //

/ — обычное деление, возвращает float.

// — деление с округлением вниз до целого числа.


print(7 / 2)    # 3.5 

print(7 // 2)   # 3 

print(-7 / 2)   # -3.5 

print(-7 // 2)  # -4

print(9 // 2)   # 4

print(-9 // 2)  # -5


При отрицательных числах // округляет в сторону минус бесконечности.

### 7. Преобразование типов

Для преобразования используются функции int(), float(), str().


a = int("10")

b = float("3.5")

c = str(25)

print(a, b, c)


Ошибка возникает, если попытаться преобразовать нечисловую строку в число, например int("Привет").

### 8. Вычисление — это выражение, возвращающее значение

Выражение выполняет действие и возвращает результат.


a = 5 + 3b = 10 > 5c = True and False

print(a)  # 8

print(b)  # True

print(c)  # False


### 9. Операторы сравнения и логические операторы

Операторы сравнения:

равно ==
не равно =!
меньше <
больше >
меньше или равно <=
больше или равно >=

Логические операторы:

and — И.
or — ИЛИ.
not — НЕ.

Примеры:


print(5 > 3 and 10 > 5)  # True

print(5 < 3 or 10 > 5)   # True

print(not 5 > 3)         # False


Сначала выполняется not, затем and, потом or.

### 10. Операции над разными типами

Python может автоматически преобразовывать некоторые типы.


a = 5b = 2.5print(a + b)  # 7.5


int автоматически преобразуется в float.

Но строку и число нельзя складывать напрямую:


print("Возраст: " + str(21))


Для этого необходимо использовать явное преобразование str().

### 11. Способы вывода нескольких значений


name = "Максим"

age = 21

print(name, age)

print(f"Имя: {name}, возраст: {age}")

print("Имя: {}, возраст: {}".format(name, age))

print("Имя: " + name + ", возраст: " + str(age))

Запятые — удобны для быстрого вывода.
f-строки — удобны для форматирования.
.format() — позволяет подставлять значения в шаблон.
Конкатенация — объединяет строки.

### 12. Порядок выполнения операций

Порядок выполнения определяет, какие действия выполняются первыми. Изменить его можно скобками.


result = (5 + 3) * 2 - 4print(result)


Сначала 5 + 3 = 8, затем 8 * 2 = 16, затем 16 - 4 = 12.

### 13. Множественное присваивание

Позволяет присваивать значения нескольким переменным одновременно.


a, b, c = 10, 20, 30x, y = 5, 10x, y = y, xnumbers = [1, 2, 3]a, b, c = numbers


Значения переменных можно менять местами без дополнительной переменной.

### 14. Изменяемые и неизменяемые типы

Изменяемые типы можно менять после создания, неизменяемые — нельзя.

Числа и строки относятся к неизменяемым типам.


text = "Привет"

print(id(text))text = "Пока"

print(id(text))


При присваивании новой строки создаётся другой объект, а исходная строка не изменяется.
